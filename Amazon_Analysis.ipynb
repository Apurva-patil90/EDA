### Outliers Detection & Handling

````python
# Identify and handle outliers using the IQR method

# Define a function to handle outliers using the IQR method
def handle_outliers_iqr(df, column):
    """
    Handles outliers in a numerical column using the IQR method.

    Args:
        df (pd.DataFrame): The DataFrame.
        column (str): The name of the column to process.

    Returns:
        pd.DataFrame: The DataFrame with outliers removed for the specified column.
    """
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    # Filter out the outliers
    df_cleaned = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)].copy()
    return df_cleaned

# Apply outlier handling to numerical columns (excluding 'seasons', 'imdb_votes', 'tmdb_popularity')
numerical_cols_to_process = ['runtime', 'release_year', 'imdb_score', 'tmdb_score']

for col in numerical_cols_to_process:
    cleaned_titles = handle_outliers_iqr(cleaned_titles, col)

# Handle rating extremes again after filtering other columns (in case some were reintroduced)
cleaned_titles = cleaned_titles[(cleaned_titles['imdb_score'] > 1) & (cleaned_titles['imdb_score'] < 10)].copy()


# Visualization

# Visualize 'runtime' distribution BEFORE outlier handling
plt.figure(figsize=(10, 5))
sns.boxplot(x=titles['runtime'])
plt.title('Runtime Distribution Before Outlier Handling')
plt.show()

# Visualize 'runtime' distribution AFTER outlier handling
plt.figure(figsize=(10, 5))
sns.boxplot(x=cleaned_titles['runtime'])
plt.title('Runtime Distribution After Outlier Handling')
plt.show()

# Visualize 'imdb_score' distribution BEFORE outlier handling
plt.figure(figsize=(10, 5))
sns.boxplot(x=titles['imdb_score'])
plt.title('IMDB Score Distribution Before Outlier Handling')
plt.show()

# Visualize 'imdb_score' distribution AFTER outlier handling
plt.figure(figsize=(10, 5))
sns.boxplot(x=cleaned_titles['imdb_score'])
plt.title('IMDB Score Distribution After Outlier Handling')
plt.show()
````
